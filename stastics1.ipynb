```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
    "Name":["A","B","c","D","E"],
    "Marks":[65,72,81,55,98]
}
df=pd.DataFrame(data)
print(df)
print(df["Marks"].mean())
plt.bar(df["Name"],df["Marks"])
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student marks")
plt.show()
```

```python
result=np.random.choice(["Head","Tail"],size=1000)
heads=np.sum(result=="Head")

print("Heads:",heads)
print("Probabilty of Head:",heads/1000)
counts=[
    np.sum(result=="Head"),
    np.sum(result=="Tail")
]
plt.bar(["Head","Tail"],counts)
plt.xlabel("Outcome")
plt.ylabel("Count")
plt.title("coin toss")
plt.show()
```

```python
import numpy as np
roll=np.random.choice([1,2,3,4,5,6])
print("you rolled roll",roll)

rolls=np.random.choice([1,2,3,4,5,6],size=10)
print("10 rolled",rolls)

rolls=np.random.choice([1,2,3,4,5,6],size=100)
print("100 rolled",rolls)

for face in range(1,7):
  count=np.sum(rolls==face)
  print(face,"was rolled",count,"times","probability",count/1000)
```

```python
rolls=np.random.randint(1,7,10)
for number in range(1,7):
  count=np.sum(rolls==number)
  print(number,"was called",count,"times","probability",count/100)
```
